#**Conversion of Tensor to Audio**

```python
import torch
import torchaudio  # to save audio file
import math  # for math calculations
```

```python
# Sample rate
# 16000 sound samples per second
# Audio is made of many tiny sound measurements.
sample_rate = 16000
```

```python
# Create Time Values
# Creates 16000 evenly spaced numbers from 0 seconds → 1 second
# 0.0000
#  0.0001
#  0.0002
#  ...
#  1.0000

t = torch.linspace(0, 1, sample_rate)
print(t)
```

```python
frequency = 440
```

```python
# creates the actual sound wave.
# Creates sine wave values like
audio_tensor = torch.sin(2 * math.pi * frequency * t)
print(audio_tensor)
```

```python
audio_tensor.shape
```

```python
# Add Channel Dimension
#Audio libraries expect shape: (channels, samples)
#Originally shape was: (16000,)

audio_tensor = audio_tensor.unsqueeze(0)
audio_tensor.shape
```

```python
# Save as Audio File

torchaudio.save("sound.wav", audio_tensor, sample_rate)
```

```python
print("Audio file saved successfully!")
```
