## **Step 1: Import the MNIST Dataset**

```python
from tensorflow.keras.datasets import mnist
```

```python
# Load dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

# 1.**List**

Syntax:

    list_name = [item1, item2, item3, ...]

### **1.1 Store a few MNIST images**

```python
image_list = [x_train[0], x_train[1], x_train[2]]

print("Number of images:", len(image_list))
```

```python
image_list[1]
```

### ***Common Methods***

### **Let's add "x_train[3] -> 1" to the list.**

```python
image_list.append(x_train[3])
print("Number of images:", len(image_list))
```

### **Insert an Item**

```python
image_list.insert(1, x_train[4])
print("Number of images:", len(image_list))
```

### **Remove an item from a list.**

```python
del image_list[3]
print("Number of images:", len(image_list))
```

## **2. Tuple**

| Feature          | List (`list`)        | Tuple (`tuple`)               |
| ---------------- | -------------------- | ----------------------------- |
| Syntax           | `[ ]`                | `( )`                         |
| Mutable          | ✅ Can be changed     | ❌ Cannot be changed        |

```python
image_info = (28, 28, "grayscale")

print(image_info)
```

### ***Common Methods***

### **Returns how many times "28" appears in the tuple.**

```python
print(image_info.count(28))
```

### **Single element tuple**

```python
image_info = (28)
type(image_info)
```

```python
image_info = (28,)
type(image_info)
```

## **4. Dictionary**

| Feature            | Set (`set`)                           | Dictionary (`dict`)                    |
| ------------------ | ------------------------------------- | -------------------------------------- |
| Stores             | Unique values only                    | Key-value pairs                        |
| Syntax             | `{1, 2, 3}`                           | {"a": 1, "b": 2}

Syntax:

    dictionary_name = { key1: value1, key2: value2, key3: value3 }

### **Stores data using key-value pairs.**

```python
sample = {
    "image": x_train[1],
    "label": y_train[1]
}

print(sample["label"])
```

## **5. Dictionary of Lists**

```python
sample = {
    "image0": x_train[0],
    "label0": y_train[0],
    "image1": x_train[1],
    "label1": y_train[1],
    "image2": x_train[2],
    "label2": y_train[2],
}
```

### **Store images and labels separately.**

```python
dataset = {
    "images": [x_train[0], x_train[1], x_train[2]],
    "labels": [y_train[0], y_train[1], y_train[2]]
}

print(dataset["labels"])
```
