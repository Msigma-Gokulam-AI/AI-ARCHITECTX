# **Load and Show an Image**

**Step 1: Import OpenCV**

```python
from google.colab.patches import cv2_imshow
import cv2 as cv
```

**Step 2: Upload an Image**

```python
from google.colab import files
uploaded = files.upload()
```

**Step 3: Read an Image**

```python
img = cv.imread("car.jpg")
```

**Step 4: Show an Image**

```python
cv2_imshow(img)
```

**Convert to grayscale using cvtcolor()**

```python
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
```

```python
cv2_imshow(gray)
```

## **Preprocessing**

### **1. Rotation**

```python
rotated = cv.rotate(img,cv.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

### **2. Flip**

```python
flip = cv.flip(img, 0)
cv2_imshow(flip)
```

### **3.Resize**

```python
resized = cv.resize(img, (350, 250))
cv2_imshow(resized)
```
