## Processing Alice's Subjects (Definite Iteration)
To calculate her total marks over a fixed number of exam **subjects**, Alice implements a **For Loop**.

```python
total_score = 0
for i in range(1, 4):
  marks_input = input("Enter marks for subject " + str(i) + ":" )
  marks = int(marks_input)

  total_score = total_score + marks

print("\nTotal Score: " , total_score)
```

```python
print("\n--- Step 1: Processing Alice's 3 Subjects ---")
total_score = 0

# Definite Iteration: Loops exactly 3 times (1, 2, and 3)
for i in range(1, 4):

  # input() captures data as a string (str)

  #To convert integer(int) to string(str) --> str(i)
  # output : Enter marks for subject 1:
  marks_input = input("Enter marks for subject " + str(i) + ":" )

  # Convert string to Integer (int) to perform math
  marks = int(marks_input)

  # Arithmetic operator (+) to add to the final total
  total_score = total_score + marks

print("\nTotal Score: " , total_score)

# Decision Logic for Alice
if total_score >= 100:
    print("Alice's Status: Pass")
else:
    print("Alice's Status: Fail")
```
