## High-Scale School Analytics (Indefinite Iteration)
Now, Alice wants to compute massive batches of records for the entire school district. Because she doesn't know the exact student count in advance, a For Loop won't work. Instead, she utilizes a **While Loop** for **Indefinite Iteration**, which repeats instructions dynamically as long as a condition stays **True**.

```python
continue_checking = "yes"
while continue_checking == "yes":
  continue_checking = input("Do you want to check another student? (yes/no): ")
  print(continue_checking)
```

```python
print("\n--- Step 2: Processing Other Students ---")

# Variable to control the indefinite iteration loop
continue_checking = "yes"
# Indefinite Iteration: Runs as long as the condition remains True
while continue_checking == "yes":
    other_total = int(input("Enter another student's total score: "))

    # Decision Logic inside the while loop
    if other_total >= 100:
        print("Student Status: Pass")
    else:
        print("Student Status: Fail")

    # CRUCIAL: Changing the variable inside the loop avoids the Infinite Loop Hazard!
    continue_checking = input("Do you want to check another student? (yes/no): ")

print("Program finished successfully!")
```
